### 1.1 Preprocessing Steps

For this competition, I worked with a dataset containing 4,800 posts. The text was stored in final_posts.json, while the labels and train/test information were provided in separate CSV files. I merged these files using post_id and separated the data into 3,360 training samples and 1,440 test samples.

The dataset contains six emotion categories: anger, disgust, fear, joy, sadness, and surprise. I noticed that the classes were not evenly distributed. Joy had the most training samples (875), while fear had only 288. I thought this imbalance might affect the model, especially for emotions with fewer examples.

For text preprocessing, I converted the posts to lowercase, removed URLs and user mentions, and cleaned up extra spaces. I kept the original text in a separate column so that I could still check it if needed. I did not remove stopwords or apply stemming because I wanted to preserve as much of the original meaning as possible, including words that might be useful for emotion classification.

I also used a stratified 80/20 split for training and validation. This helped keep the emotion distribution similar in both sets.

### 1.2 Feature Engineering Steps

I started with TF-IDF because it was one of the methods we had already practiced in the Master notebook. My first approach used word-level TF-IDF with unigrams and bigrams. I set min_df=2 to filter out very rare terms and enabled sublinear_tf to reduce the influence of repeated words.

The first results were not very good, so I tried several different settings. Surprisingly, using only unigrams performed better than combining unigrams and bigrams in cross-validation.

I then tested character-level TF-IDF with character n-grams from 2 to 5. This seemed useful for social media posts, since people often use abbreviations, informal expressions, and different spellings. The character-based model performed better on my validation set.

After that, I tried combining word-level and character-level TF-IDF using FeatureUnion. This allowed the model to learn from both complete words and smaller character patterns. Among the settings I tested, this combination gave me the best validation result.

### 1.3 Explanation of Your Model

I then combined word and character TF-IDF features using FeatureUnion. The best configuration used balanced class weights and C=0.3, reaching a validation Macro F1 of 0.5254. I submitted this model to Kaggle and received a Public Score of 0.5078.

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
from pathlib import Path
import pandas as pd

# Check available datasets and files
print("Current directory:", Path.cwd())

for folder in ["data", "data/semeval", "lab2_workspace"]:
    path = Path(folder)
    print(f"\n--- {folder} ---")

    if path.exists():
        for file in sorted(path.rglob("*")):
            if file.is_file():
                print(file)
    else:
        print("Folder not found")
```

```python
import pandas as pd
import json
from pathlib import Path

# Load Kaggle competition datasets
data_dir = Path("kaggle_data")

posts = pd.read_json(data_dir / "final_posts.json")
emotions = pd.read_csv(data_dir / "emotion.csv")
data_id = pd.read_csv(data_dir / "data_identification.csv")
sample = pd.read_csv(data_dir / "sampleSubmission.csv")

# Inspect the dataset structure
for name, df in [
    ("Posts", posts),
    ("Emotions", emotions),
    ("Data Identification", data_id),
    ("Sample Submission", sample)
]:
    print(f"\n--- {name} ---")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print(df.head(3))
```

```python
import re

# Merge post texts with train/test identification
df = posts.merge(
    data_id,
    on="post_id",
    how="inner",
    validate="one_to_one"
)

# Split train and test data
train_df = df[df["set"] == "train"].copy()
test_df = df[df["set"] == "test"].copy()

# Add labels to training data only
train_df = train_df.merge(
    emotions,
    on="post_id",
    how="left",
    validate="one_to_one"
)

# Basic text preprocessing
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

train_df["clean"] = train_df["text"].apply(clean_text)
test_df["clean"] = test_df["text"].apply(clean_text)

# Validate data
assert len(train_df) == 3360
assert len(test_df) == 1440
assert train_df["label"].notna().all()
assert test_df["post_id"].is_unique

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)

print("\nEmotion distribution:")
print(train_df["label"].value_counts())

print("\nSample training data:")
print(train_df[["post_id", "clean", "label"]].head())
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# Split labeled data into training and validation sets
X_train_text, X_val_text, y_train, y_val = train_test_split(
    train_df["clean"],
    train_df["label"],
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

# Convert text into TF-IDF features
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)

X_train = vectorizer.fit_transform(X_train_text)
X_val = vectorizer.transform(X_val_text)

print("Training features:", X_train.shape)
print("Validation features:", X_val.shape)
print("Vocabulary size:", len(vectorizer.vocabulary_))
```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Train a Linear SVM classifier
model = LinearSVC(
    C=1.0,
    random_state=42
)

model.fit(X_train, y_train)

# Evaluate on validation data
y_pred = model.predict(X_val)

print("Accuracy:", round(accuracy_score(y_val, y_pred), 4))
print("Macro F1:", round(f1_score(y_val, y_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(y_val, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred, labels=model.classes_))
print("Class order:", model.classes_)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
import pandas as pd

# Compare TF-IDF and Linear SVM settings using cross-validation
configs = [
    {"name": "Unigram_C1", "ngram": (1, 1), "C": 1.0, "balance": None},
    {"name": "Bigram_C1", "ngram": (1, 2), "C": 1.0, "balance": None},
    {"name": "Bigram_C0.1", "ngram": (1, 2), "C": 0.1, "balance": None},
    {"name": "Bigram_C10", "ngram": (1, 2), "C": 10.0, "balance": None},
    {"name": "Bigram_Balanced", "ngram": (1, 2), "C": 1.0, "balance": "balanced"},
    {"name": "CharNgram_Balanced", "ngram": (2, 5), "C": 1.0, "balance": "balanced"}
]

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

results = []

for cfg in configs:

    vectorizer_params = {
        "ngram_range": cfg["ngram"],
        "min_df": 2,
        "sublinear_tf": True
    }

    if cfg["name"] == "CharNgram_Balanced":
        vectorizer_params["analyzer"] = "char"

    pipeline = Pipeline([
        ("tfidf", TfidfVectorizer(**vectorizer_params)),
        ("svm", LinearSVC(
            C=cfg["C"],
            class_weight=cfg["balance"],
            random_state=42
        ))
    ])

    scores = cross_val_score(
        pipeline,
        X_train_text,
        y_train,
        cv=cv,
        scoring="f1_macro",
        n_jobs=-1
    )

    results.append({
        "Model": cfg["name"],
        "CV_Macro_F1": scores.mean(),
        "Std": scores.std()
    })

results_df = pd.DataFrame(results).sort_values(
    "CV_Macro_F1",
    ascending=False
)

display(results_df)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Best model from cross-validation
best_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        analyzer="char",
        ngram_range=(2, 5),
        min_df=2,
        sublinear_tf=True
    )),
    ("svm", LinearSVC(
        C=1.0,
        class_weight="balanced",
        random_state=42
    ))
])

# Train on the training portion only
best_pipeline.fit(X_train_text, y_train)

# Evaluate on the held-out validation set
best_pred = best_pipeline.predict(X_val_text)

best_accuracy = accuracy_score(y_val, best_pred)
best_f1 = f1_score(y_val, best_pred, average="macro")

print("Character n-gram Accuracy:", round(best_accuracy, 4))
print("Character n-gram Macro F1:", round(best_f1, 4))

print("\nClassification Report:")
print(classification_report(y_val, best_pred))

print("\nComparison:")
print("Original Bigram Macro F1: 0.4855")
print("Character n-gram Macro F1:", round(best_f1, 4))
```

```python
from sklearn.pipeline import Pipeline, FeatureUnion
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import f1_score, accuracy_score
import pandas as pd

# Try combining word-level and character-level TF-IDF
experiments = [
    {"name": "WordChar_C0.3", "C": 0.3},
    {"name": "WordChar_C1", "C": 1.0},
    {"name": "WordChar_C3", "C": 3.0},
]

results = []
candidate_models = {}

for exp in experiments:
    features = FeatureUnion([
        ("word", TfidfVectorizer(
            analyzer="word",
            ngram_range=(1, 2),
            min_df=2,
            sublinear_tf=True
        )),
        ("char", TfidfVectorizer(
            analyzer="char",
            ngram_range=(2, 5),
            min_df=2,
            sublinear_tf=True
        ))
    ])

    pipeline = Pipeline([
        ("features", features),
        ("svm", LinearSVC(
            C=exp["C"],
            class_weight="balanced",
            random_state=42
        ))
    ])

    pipeline.fit(X_train_text, y_train)
    predictions = pipeline.predict(X_val_text)

    score = f1_score(y_val, predictions, average="macro")
    accuracy = accuracy_score(y_val, predictions)

    results.append({
        "Model": exp["name"],
        "Macro_F1": round(score, 4),
        "Accuracy": round(accuracy, 4)
    })

    candidate_models[exp["name"]] = pipeline

results_df_improved = pd.DataFrame(results).sort_values(
    "Macro_F1",
    ascending=False
)

display(results_df_improved)

print("Current baseline Macro F1:", round(best_f1, 4))
```

```python
# Final competition model: pretrained DistilRoBERTa
# Evaluated on Colab using a T4 GPU.
# The model was used without fine-tuning.

import torch
import numpy as np
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.metrics import accuracy_score, f1_score

MODEL_NAME = "j-hartmann/emotion-english-distilroberta-base"
device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
transformer_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
).to(device)
transformer_model.eval()

emotion_labels = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

label_ids = [
    transformer_model.config.label2id[label]
    for label in emotion_labels
]

def transformer_predict(texts, batch_size=32):
    predictions = []

    for start in range(0, len(texts), batch_size):
        batch = texts[start:start + batch_size]

        inputs = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=128,
            return_tensors="pt"
        ).to(device)

        with torch.inference_mode():
            logits = transformer_model(**inputs).logits

        # Exclude the neutral category
        logits = logits[:, label_ids]
        indices = logits.argmax(dim=1).cpu().tolist()
        predictions.extend(emotion_labels[i] for i in indices)

    return np.array(predictions)
```

```python
# DistilRoBERTa validation
# Run this cell in Colab with the model loaded.

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report
import numpy as np

_, val_indices = train_test_split(
    np.arange(len(train_df)),
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

val_df = train_df.iloc[val_indices].copy()

val_predictions = transformer_predict(
    val_df["text"].fillna("").astype(str).tolist()
)

print("Accuracy:", round(
    accuracy_score(val_df["label"], val_predictions), 4
))

print("Macro F1:", round(
    f1_score(val_df["label"], val_predictions, average="macro"), 4
))

print(classification_report(val_df["label"], val_predictions))
```
